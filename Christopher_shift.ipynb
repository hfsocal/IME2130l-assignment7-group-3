```python
units = [410, 452, 438, 465, 480]         # units produced Monday to Friday on your shift
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']   # labels for the bars
total = sum(units)                        # weekly total
print(f'{member_name}, shift {shift_number}: {total} units this week')   # one summary line
```

```python
import matplotlib.pyplot as plt           # the plotting library

plt.bar(days, units)                      # one bar per day
plt.title(f'Shift {shift_number} output, {member_name}')   # title names the shift and the person
plt.ylabel('Units')                       # y axis label
plt.show()                                # display the chart
```
